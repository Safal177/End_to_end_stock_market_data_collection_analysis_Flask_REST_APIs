## Building a function to get company information using Yahoo Finance

```python
def extract_company_information(symbol):
    yf_ticker = yf.Ticker(symbol.upper())
    company_information = yf_ticker.info
    if not company_information or not company_information.get('longName'):
        return None
    company_data = {'company_symbol': yf_ticker.ticker,
                    'official_company_name': company_information.get('longName'),
                    'industry_classification': company_information.get('industry'),
                    'company_sector': company_information.get('sector'),
                    'country_of_registration': company_information.get('country'),
                    'company_website': company_information.get('website'),
                    'company_business_summary': company_information.get('longBusinessSummary'),
                    'key_officers': [{'name': officer.get('name'), 'title': officer.get('title')} for officer in company_information.get('companyOfficers', [])
                                    ]
                   }
    return company_data
```

```python
# Load Flask and Yahoo Finance libraries
from flask import Flask, jsonify, request
import yfinance as yf


# Initialize the Flask application for historical stock market data analysis
app = Flask(__name__)



# Building a REST API GET endpoint for company information using Flask
@app.route('/api/company/<symbol>', methods = ['GET'])
def get_company(symbol):
    company_data = extract_company_information(symbol)
    # If true, go this route.
    if company_data is None:
        return jsonify({'company_information_error': 'I could not find data about this Company'}), 404
    #  If the above condition is "False", return the information stored in company_data.
    return jsonify(company_data), 200

# Building a REST API GET endpoint for Stock Market Data using Flask
@app.route('/api/stock/<symbol>', methods = ['GET'])
def get_stock(symbol):
    stock_data = extract_stock_market_data(symbol)
    # If true, go this route.
    if stock_data is None:
        return jsonify({'stock_market_data_error': 'I could not find Stock Market Data.'}), 404
    return jsonify(stock_data), 200


# Building a REST API POST endpoint for the historical market Stock Data  using Flask
@app.route('/api/historical', methods = ['POST'])
def get_historical():
    data = request.get_json(silent=True) or {}
    symbol = data.get('symbol')
    initial_date = data.get('initial_date')
    final_date = data.get('final_date')
    # If true, go this route.
    if not symbol or not initial_date or not final_date:
        return jsonify({'missing key parameters': 'Stock symbol, initial date, and final date are missing'}), 400
    historical_data = extract_historical_stock_market_data(symbol, initial_date, final_date)
    # If true, go this route.
    if historical_data is None:
        return jsonify({'status': 'No historical stock market data is seen.'}), 404
    return jsonify(historical_data), 200


# Building a REST API POST endpoint for historical stock market analytical insights using Flask
@app.route('/api/analytics', methods = ['POST'])
def get_analytical_insights():
    data = request.get_json(silent=True) or {}
    symbol = data.get('symbol')
    initial_date = data.get('initial_date')
    final_date = data.get('final_date')
    if not symbol or not initial_date or not final_date:
        return jsonify({'missing key parameters': 'Stock symbol, initial date, and final date are missing'}), 400
    analytical_insights = extract_analytical_insights(symbol, initial_date, final_date)
    if analytical_insights is None:
        return jsonify({'failure_reason': 'For analytical insights, historical stock market data could not be found.'}), 404
    
    #  If the above condition is "False", return the information stored in analytical_insights.
    return jsonify(analytical_insights), 200





# Building a function to get company information using Yahoo Finance

def extract_company_information(symbol):
    yf_ticker = yf.Ticker(symbol.upper())
    company_information = yf_ticker.info
    # If true, go this route.
    if not company_information or not company_information.get('longName'):
        return None
    
    company_data = { 'company_symbol': yf_ticker.ticker,
                    'official_company_name': company_information.get('longName'),
                    'industry_classification': company_information.get('industry'),
                    'company_sector': company_information.get('sector'),
                    'country_of_registration': company_information.get('country'),
                    'company_website': company_information.get('website'),
                    'company_business_summary': company_information.get('longBusinessSummary'),
                    'key_officers': [{'name': officer.get('name'), 'title': officer.get('title')} for officer in company_information.get('companyOfficers', [])
                                     ]
                                     }
    return company_data


#  Building a function to get stock market data using Yahoo Finance
def extract_stock_market_data(symbol):
    yf_ticker = yf.Ticker(symbol.upper())
    current_stock_data = yf_ticker.info
    current_price = current_stock_data.get('currentPrice')
    previous_close = current_stock_data.get('previousClose')
    if current_price is None or previous_close is None or previous_close == 0:
        return None
    stock_price_change_dollars = current_price - previous_close
    stock_price_change_percentage = (stock_price_change_dollars / previous_close) * 100
    stock_market_data = {'company_stock_code': yf_ticker.ticker,
                         'opening_stock_price': current_stock_data.get('open'),
                         'previous_day_closing_price': previous_close,
                         'current_share_price': current_price,
                         'stock_currency': current_stock_data.get('currency'),
                         'stock_exchange': current_stock_data.get('exchange'),
                         'daily_trading_volume': current_stock_data.get('volume'),
                         'highest_stock_price_today': current_stock_data.get('dayHigh'),
                         'fifty_two_week_highest_price': current_stock_data.get('fiftyTwoWeekHigh'),
                         'average_trading_volume': current_stock_data.get('averageVolume'),
                         'lowest_stock_price_today': current_stock_data.get('dayLow'),
                         'fifty_two_week_lowest_price': current_stock_data.get('fiftyTwoWeekLow'),
                         'earnings_per_share': current_stock_data.get('trailingEps'),
                         'market_operating_status': current_stock_data.get('marketState'),
                         'stock_price_change_dollars': round(stock_price_change_dollars, 3),
                         'stock_price_change_percentage': round(stock_price_change_percentage, 3)
                         }
    return stock_market_data


# Building a function to get historical stock market data using Yahoo Finance
def extract_historical_stock_market_data(symbol, initial_date, final_date):
    yf_ticker = yf.Ticker(symbol.upper())
    historical_stock_market_data = yf_ticker.history(start = initial_date, end = final_date)
    if historical_stock_market_data.empty:
        return None
    historical_stock_market_data = historical_stock_market_data.reset_index()
    historical_stock_market_data['Date'] = historical_stock_market_data['Date'].dt.strftime('%Y-%m-%d')
    historical_stock_market_data_records = historical_stock_market_data.to_dict(orient = 'records')
    
    return historical_stock_market_data_records

# Developing a function to analyze historical stock market data
def extract_analytical_insights(symbol, initial_date, final_date):
    yf_ticker = yf.Ticker(symbol.upper())
    historical_stock_market_data = yf_ticker.history(start = initial_date,
                                                     end = final_date
                                                    )
    if historical_stock_market_data.empty:
        return None
    initial_closing_price = historical_stock_market_data['Close'].iloc[0]
    final_closing_price = historical_stock_market_data['Close'].iloc[-1]
    historical_stock_price_change = final_closing_price - initial_closing_price
    historical_stock_price_change_percentage = (historical_stock_price_change / initial_closing_price) * 100
    minimum_closing_price = historical_stock_market_data['Close'].min()
    maximum_closing_price = historical_stock_market_data['Close'].max()
    average_closing_price = historical_stock_market_data['Close'].mean()
    median_closing_price = historical_stock_market_data['Close'].median()
    closing_price_standard_deviation = historical_stock_market_data['Close'].std()
    minimum_volume = historical_stock_market_data['Volume'].min()
    maximum_volume = historical_stock_market_data['Volume'].max()
    mean_volume = historical_stock_market_data['Volume'].mean()
    total_historical_volume = historical_stock_market_data['Volume'].sum()
    
    analytical_insights = {'company_stock_code': yf_ticker.ticker,
                           'initial_date': initial_date,
                           'final_date': final_date,
                           'number_of_trading_days': len(historical_stock_market_data),
                           'initial_closing_price': round(float(initial_closing_price), 3),
                           'final_closing_price': round(float(final_closing_price), 3),
                           'historical_stock_price_change': round(float(historical_stock_price_change), 3),
                           'historical_stock_price_change_percentage': round(float(historical_stock_price_change_percentage), 3),
                           'minimum_closing_price': round(float(minimum_closing_price), 3),
                           'maximum_closing_price': round(float(maximum_closing_price), 3),
                           'average_closing_price': round(float(average_closing_price), 3),
                           'median_closing_price': round(float(median_closing_price), 3),
                           'closing_price_standard_deviation': round(float(closing_price_standard_deviation), 3),
                           'minimum_volume': int(minimum_volume),
                           'maximum_volume': int(maximum_volume),
                           'mean_volume': round(float(mean_volume), 3),
                           'total_historical_volume': int(total_historical_volume)
                          }
    return analytical_insights



# Start the Flask REST API server
if __name__ == '__main__':
    app.run(debug=True)
```
